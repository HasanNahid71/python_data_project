```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
pd.read_csv("https://lukeb.co/software_csv")
```

```python
df_US_software_pivot = pd.read_csv("https://lukeb.co/software_csv", index_col='job_posted_month')

df_US_software_pivot
```

```python
df_US_software_pivot.merge(df_US_software_pivot, on="job_posted_month")
```

```python
df_US_merged=df_US_software_pivot.merge(df_US_software_pivot, on="job_posted_month")
```

```python
top_5 = df_US_merged.sum()

top_5
```

```python
top_5 = df_US_merged.sum().sort_values(ascending=False)

top_5
```

```python
top_5 = df_US_merged.sum().sort_values(ascending=False).head()

top_5
```

```python
top_5 = (
    df_US_merged
    .sum()
    .sort_values(ascending=False)
    .head()
    .index
    .to_list()
)

top_5
```

```python
df_US_merged[top_5].plot(kind='line')
```

```python

df_US_merged[top_5].plot(kind='line')
plt.title('Monthly Job Postings for Top Tech Jobs in the US')
plt.xlabel('2023')
plt.ylabel('Job Count')
plt.ylim(0, 20000)
plt.legend()
plt.show()
```
